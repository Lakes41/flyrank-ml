## 3. Quick look at the data (2-3 real numbers)

Out of 30,000 pages in the starter dataset, 16,262 pages (54.2%) are trending downwards. However, if we filter for pages that are both declining AND have had significant recent visibility (over 500 impressions in the last 90 days), we find a smaller, highly prioritized group of 9,956 pages to focus our recovery efforts on.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

total_pages = len(df)
declining_pages = len(df[df['trend_direction'] == 'down'])
declining_visible = len(df[(df['trend_direction'] == 'down') & (df['impressions_90d'] > 500)])

print(f"Total pages in dataset: {total_pages:,}")
print(f"Pages trending down: {declining_pages:,} ({declining_pages/total_pages:.1%})")
print(f"Declining AND high visibility (>500 imp): {declining_visible:,}")
```
